```python
import numpy as np
from gensim.models import KeyedVectors
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense
from tensorflow.keras.models import Sequential
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.datasets import imdb
```

```python
max_features = 10000
(X_train, y_train), (X_test, y_test) = imdb.load_data(num_words=max_features)
word_index = imdb.get_word_index()
```

```python
word2vec = KeyedVectors.load_word2vec_format('../embedding-word2vec/GoogleNews-vectors-negative300.bin', binary=True)
```

```python
embedding_dim = 300
embedding_matrix = np.zeros((max_features, embedding_dim))
for word, i in word_index.items():
    if i < max_features and word in word2vec:
        #if word in word2vec:
        embedding_vector = word2vec[word]
        #if embedding_vector is not None:
        embedding_matrix[i] = embedding_vector
```

```python
max_len = 500
X_train = pad_sequences(X_train, maxlen=max_len)
X_test = pad_sequences(X_test, maxlen=max_len)
```

```python
model = Sequential()
model.add(Embedding(input_dim=max_features, output_dim=embedding_dim, 
                    weights=[embedding_matrix], input_length=max_len, trainable=False))
model.add(SimpleRNN(128, activation='relu'))
model.add(Dense(1, activation='sigmoid'))

model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
from tensorflow.keras.callbacks import EarlyStopping
earlystopping=EarlyStopping(monitor='val_loss',patience=5,restore_best_weights=True)
earlystopping
```

```python
history=model.fit(
    X_train,y_train,epochs=10,batch_size=32,
    validation_split=0.2,
    callbacks=[earlystopping]
)
```

```python
model.evaluate(X_test, y_test)
```
