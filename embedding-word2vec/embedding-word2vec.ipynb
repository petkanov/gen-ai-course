```python
from gensim.models import KeyedVectors
from tensorflow.keras.layers import Embedding
from tensorflow.keras.utils import pad_sequences
from tensorflow.keras.models import Sequential
import numpy as np
```

```python
word2vec_path = 'GoogleNews-vectors-negative300.bin'  # Ensure you have this file
word2vec = KeyedVectors.load_word2vec_format(word2vec_path, binary=True)
```

```python
dim = 300
sent = [
    'the glass of milk',
    'the glass of juice',
    'the cup of tea',
    'I am a good boy',
    'I am a good developer',
    'understand the meaning of words',
    'your videos are good',
]
sent_length = 8
```

```python
tokenizer = {}
index = 1
for sentence in sent:
    for word in sentence.split():
        if word not in tokenizer:
            tokenizer[word] = index
            index += 1
```

```python
tokenizer
```

```python
embedding_matrix = np.zeros((len(tokenizer) + 1, dim))
for word, i in tokenizer.items():
    if word in word2vec:
        embedding_matrix[i] = word2vec[word]
```

```python
print(embedding_matrix.shape)
embedding_matrix
```

```python
one_hot_repr = [[tokenizer[word] for word in sentence.split()] for sentence in sent]
embedded_docs = pad_sequences(one_hot_repr, padding='pre', maxlen=sent_length)
```

```python
model = Sequential()

model.add(Embedding(input_dim=len(tokenizer) + 1, output_dim=dim, weights=[embedding_matrix], trainable=False))
```
