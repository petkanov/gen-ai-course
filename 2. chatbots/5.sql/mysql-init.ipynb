# Complex SCHOOL multy-table Database 

```python
import mysql.connector

# Connect to MySQL server
connection = mysql.connector.connect(
    host="localhost",  # Your MySQL server address
    user="root",       # Your MySQL username
    password="admin"  # Your MySQL password
)

cursor = connection.cursor()

cursor.execute("DROP DATABASE IF EXISTS _db")

# Create a new database
cursor.execute("CREATE DATABASE IF NOT EXISTS school_db")
cursor.execute("USE school_db")
```

#### Create the STUDENT, DISCIPLINE and DISCIPLINE_JOINT tables

```python
# Create the STUDENT table
student_table = """
CREATE TABLE IF NOT EXISTS STUDENT (
    STUDENT_ID INT AUTO_INCREMENT PRIMARY KEY,
    NAME VARCHAR(50) NOT NULL,
    CLASS_LEVEL VARCHAR(25),  -- e.g., Freshman, Sophomore
    ENROLLED_CLASSES INT DEFAULT 0,
    MONEY_SPENT DECIMAL(10, 2) DEFAULT 0.0,  -- For additional lessons
    HOURS_MISSED INT DEFAULT 0,
    DATE_OF_BIRTH DATE,
    CONTACT_EMAIL VARCHAR(100),
    CREATED_AT TIMESTAMP DEFAULT CURRENT_TIMESTAMP
)
"""
cursor.execute(student_table)

# Create the DISCIPLINE table
discipline_table = """
CREATE TABLE IF NOT EXISTS DISCIPLINE (
    DISCIPLINE_ID INT AUTO_INCREMENT PRIMARY KEY,
    NAME VARCHAR(50) NOT NULL UNIQUE,  -- e.g., Mathematics, Science
    DESCRIPTION TEXT,
    CREDITS INT DEFAULT 0,
    CREATED_AT TIMESTAMP DEFAULT CURRENT_TIMESTAMP
)
"""
cursor.execute(discipline_table)

# Create the DISCIPLINE_JOINT table for many-to-many relationship
discipline_joint_table = """
CREATE TABLE IF NOT EXISTS DISCIPLINE_JOINT (
    STUDENT_ID INT,
    DISCIPLINE_ID INT,
    ENROLL_DATE TIMESTAMP DEFAULT CURRENT_TIMESTAMP,
    PERFORMANCE_GRADE CHAR(2),  -- Grade e.g., A, B, C
    PRIMARY KEY (STUDENT_ID, DISCIPLINE_ID),
    FOREIGN KEY (STUDENT_ID) REFERENCES STUDENT(STUDENT_ID) ON DELETE CASCADE,
    FOREIGN KEY (DISCIPLINE_ID) REFERENCES DISCIPLINE(DISCIPLINE_ID) ON DELETE CASCADE
)
"""
cursor.execute(discipline_joint_table)
```

#### Inserting records into tables

```python
# Insert records into STUDENT table
students = """
INSERT INTO STUDENT (NAME, CLASS_LEVEL, ENROLLED_CLASSES, MONEY_SPENT, HOURS_MISSED, DATE_OF_BIRTH, CONTACT_EMAIL)
VALUES (%s, %s, %s, %s, %s, %s, %s)
"""
student_records = [
    ('Krish Patel', 'Junior', 6, 1500.50, 4, '2005-03-15', 'krish@example.com'),
    ('John Doe', 'Sophomore', 5, 1250.00, 8, '2006-07-20', 'john.doe@example.com'),
    ('Mukesh Sharma', 'Senior', 7, 2000.75, 2, '2004-11-12', 'mukesh@example.com'),
    ('Jacob Hill', 'Freshman', 4, 800.00, 15, '2007-01-25', 'jacob@example.com'),
    ('Dipesh Verma', 'Freshman', 3, 500.00, 20, '2007-05-30', 'dipesh@example.com'),
]

cursor.executemany(students, student_records)

# Insert records into DISCIPLINE table
disciplines = """
INSERT INTO DISCIPLINE (NAME, DESCRIPTION, CREDITS)
VALUES (%s, %s, %s)
"""
discipline_records = [
    ('Mathematics', 'Study of numbers, shapes, and patterns', 3),
    ('Science', 'Study of the physical and natural world', 4),
    ('History', 'Study of past events', 2),
    ('English', 'Study of language and literature', 3),
    ('Physical Education', 'Physical training and exercise', 1),
]

cursor.executemany(disciplines, discipline_records)

# Insert records into DISCIPLINE_JOINT table
discipline_joint = """
INSERT INTO DISCIPLINE_JOINT (STUDENT_ID, DISCIPLINE_ID, PERFORMANCE_GRADE)
VALUES (%s, %s, %s)
"""
discipline_joint_records = [
    (1, 1, 'A'),  # Krish enrolled in Mathematics
    (1, 2, 'B'),  # Krish enrolled in Science
    (2, 1, 'A'),  # John enrolled in Mathematics
    (2, 3, 'B'),  # John enrolled in History
    (3, 4, 'A'),  # Mukesh enrolled in English
    (3, 5, 'C'),  # Mukesh enrolled in Physical Education
    (4, 2, 'C'),  # Jacob enrolled in Science
    (4, 3, 'B'),  # Jacob enrolled in History
    (5, 5, 'A'),  # Dipesh enrolled in Physical Education
]

cursor.executemany(discipline_joint, discipline_joint_records)
```

#### Fetch and display data for verification

```python
cursor.execute("SELECT * FROM STUDENT")
print("STUDENT Table Records:")
for row in cursor.fetchall():
    print(row)

cursor.execute("SELECT * FROM DISCIPLINE")
print("\nDISCIPLINE Table Records:")
for row in cursor.fetchall():
    print(row)

cursor.execute("SELECT * FROM DISCIPLINE_JOINT")
print("\nDISCIPLINE_JOINT Table Records:")
for row in cursor.fetchall():
    print(row)
```

#### Commit changes and close the connection

```python
connection.commit()
connection.close()
```
